```python
import cv2
import matplotlib.pyplot as plt
import os

# Correct path
images_path = './dataset/images/'

# List all image files (only jpg/jpeg/png)
image_files = [f for f in os.listdir(images_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
print(f"Found {len(image_files)} images.")

# Pick one image
image_path = os.path.join(images_path, image_files[0])

# Load the image using OpenCV
image = cv2.imread(image_path)
# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Show the image
plt.figure(figsize=(8, 8))
plt.imshow(image)
plt.axis('off')
plt.title('Sample Image from DeepFashion')
plt.show()
```

```python
# Code Chunk 2: Print image shape
print(f"Image shape (Height, Width, Channels): {image.shape}")
```

```python
# Code Chunk 3: Show random image
import random

random_image_path = os.path.join(images_path, random.choice(image_files))
random_image = cv2.imread(random_image_path)
random_image = cv2.cvtColor(random_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 8))
plt.imshow(random_image)
plt.axis('off')
plt.title('Random Sample Image from DeepFashion')
plt.show()
```
